## Análises de Dados de Exemplo
Aqui vamos carregar e explorar os dados de comparação de preços.


```python
# Mock data para demonstração
np.random.seed(42)

# Cria dados de preços simulados
n_observacoes = 100

dados_preco = pd.DataFrame({
    'ID_Produto': np.arange(1, n_observacoes + 1),
         'Categoria': np.random.choice(['Eletrônicos', 'Casa', 'Vestuário', 'Alimentos'], n_observacoes),
         'Preco_Atual': np.random.normal(100, 20, n_observacoes).clip(20, 300),
         'Preco_Concorrencia': np.random.normal(95, 18, n_observacoes).clip(20, 300),
         'Mes': np.random.randint(1, 13, n_observacoes),
         'Ano': 2024
})

# Exibe os primeiros dados
display(dados_preco.head(10))

# Estatísticas descritivas
display(dados_preco[['Preco_Atual', 'Preco_Concorrencia']].describe())
```
